```python
df5 = spark.read.parquet("hdfs://namenode:9000/project-data//part-00000-58538248-1e4d-437a-8ecf-1e8ad91ba4f4-c000.snappy.parquet")

print(f"rows: {df5.count():,}")
print(f"columns: {len(df5.columns)}")
df5.printSchema()
```

```python
df5.limit(5).toPandas()
```

```python
from pyspark.sql.functions import hour

df5 = df5.withColumn("pickup_hour", hour("tpep_pickup_datetime"))
df5.select("tpep_pickup_datetime", "pickup_hour").show(5)
```

```python
df5.groupBy("pickup_hour")\
.count()\
.orderBy("pickup_hour")\
.show(24)
```

```python
df5 = df5.withColumn("dropoff_hour", hour("tpep_dropoff_datetime"))
df5.select("tpep_dropoff_datetime", "dropoff_hour").show(5)
```

```python
df5.groupBy("dropoff_hour")\
.count()\
.orderBy("dropoff_hour")\
.show(24)
```

```python
from pyspark.sql.functions import unix_timestamp, round as spark_round

df5 = df5.withColumn(
    "trip_duration_minutes",
    spark_round(
        (unix_timestamp("tpep_dropoff_datetime") - 
         unix_timestamp("tpep_pickup_datetime")) / 60.0, 
        2
    )
)
df5.select(
    "tpep_pickup_datetime",
    "tpep_dropoff_datetime",
    "trip_duration_minutes"
).show(5)
```

```python
from pyspark.sql.functions import dayofweek

df5 = df5.withColumn("day_of_week", dayofweek("tpep_pickup_datetime"))
df5.select("tpep_pickup_datetime", "day_of_week").show(5)
```

```python
df5.groupBy("day_of_week")\
.count()\
.orderBy("day_of_week")\
.show(7)
```

```python
from pyspark.sql.functions import col

selected_columns = [
    "VendorID",
    "passenger_count",
    "trip_distance",
    "pickup_hour",
    "day_of_week",
    "trip_duration_minutes",
    "RatecodeID",
    "payment_type",
    "fare_amount",
    "PULocationID",
    "DOLocationID",
    "payment_type_name",
    "Ratecode_name",
    "total_amount"
]
df_engineered = df5.select([col(c) for c in selected_columns])
print(f"final_shape: {df_engineered.count():,} rows x {len(df_engineered.columns)} columns")
df_engineered.printSchema()
```

```python
rr=df_engineered.describe()
rr.toPandas()
```
